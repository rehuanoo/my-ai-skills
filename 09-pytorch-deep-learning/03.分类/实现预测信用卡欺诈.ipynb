```python
from torch import nn
import torch
import pandas as pd
import numpy as np


data = pd.read_csv('./data.csv')
data.head()
```

```python
Y = data.iloc[:,14]
Y
```

```python
X = data.iloc[:,:-1]
X.head()
```

```python
# 分组的方式，来获取是否有信用欺诈，分别有多少人
Y.value_counts()
```

```python
X.shape
```

```python
# 将数据转换为张量
X= torch.tensor(X.values,dtype=torch.float32)
Y=torch.tensor(Y.values,dtype=torch.float32).reshape(-1,1)
# print(X)
# Y
```

```python
# 定义模型
model = nn.Sequential(
    # 定义一个线性层，14维度输入，1维度输出
    nn.Linear(14,1),
    # 定义激活函数
    nn.Sigmoid()
)
model
```

```python
# 定义损失函数（二分类函数）
loss_fn = nn.BCELoss()
# 定义优化器，将损失函数降低，误差降低，使用随机梯度下降方法
opt = torch.optim.SGD(model.parameters(),lr=0.00001)
```

```python
batches = 20
# 定义训练函数
for epoch in range(2000):
    for batch in range(620//20):
        # 每个批次开始的索引值
        start = batch*batches
        # 每个批次结束的位置
        end = start + batches
        x = X[start:end]
        y = Y[start:end]
        # 计算每个批次的预测值
        y_pred = model(x)
        # 计算每个批次的损失
        loss = loss_fn(y_pred,y)
        # 梯度归零
        opt.zero_grad()
        # 反向传播
        loss.backward()
        # 使用优化器跟新参数
        opt.step()
```

```python
# 看看训练预测的准确率
((model(X).data.numpy()>0.5).astype('int')==Y.numpy()).mean()
```

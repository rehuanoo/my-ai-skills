```python
import torch

A = torch.arange(20).reshape(4,5)
A
```

```python
# 转置（行变列，列变行）
A.T
```

```python
# 对称矩阵
B= torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
B
```

```python
B.T
```

```python
# 矩阵matrix 向量vector积
A = torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
x=torch.tensor([1,2,2])
torch.mv(A,x)
```

```python
# 矩阵 矩阵积
A = torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
x=torch.tensor([[1,2,2],[22,11,33],[0,2,5]])
# 行数要匹配（有几组要一样）
torch.mm(A,x)
```

```python
# 范数>>>表示向量/矩阵有多大
# 范数L2
a = torch.tensor([1.0,1])
torch.norm(a)
```

```python
# 范数L1
torch.abs(a).sum()
```

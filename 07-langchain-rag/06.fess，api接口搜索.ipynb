```python
import requests
import json

kw = '华为'
url=f'http://localhost:8080/api/v1/documents?q={kw}&num=10'

response = requests.get(url)

if response.status_code == 200:
    jsonData = response.json()
    # print(jsonData)
    for item in jsonData["data"]:
        title = item["title"]
        digest_text = item["digest"]  # 你要的那段文本
        file_path = item["site_path"]
        print(f"文档标题：{title}")
        print(f"摘要内容：{digest_text}")
        print(f"文件路径：{file_path}")
        print("-" * 50)
    # jsonData = json.loads(response.text)
    # print(jsonData)
```

```python
import os
from openai import OpenAI


def get_chat(prompt):
    messages=[{"role": "user", "content":prompt}]
    
    client = OpenAI(
    base_url="https://token.sensenova.cn/v1",
    # 若没有配置环境变量，请将下行替换为：api_key="sk-xxx"
    api_key=os.getenv('SENSENOVA_API_KEY', ''),
    )
    
    resp = client.chat.completions.create(
        model="sensenova-6.8-flash-lite",
        messages=messages,
        # temperature = 0.7
        stream = True
    )

    for chunk in resp:
        # delta = chunk.choices[0].delta
        # print(delta.content, end='')
        
        # print(chunk.choices[0].delta.content,end='')
        print(chunk)
        # print(chunk.choices[0].delta.reasoning,end='')
        

    # return resp.choices[0].message.content


```

作为商汤科技自主研发的日日新融合模态大模型（SenseNova），我的技术架构遵循商汤在多模态理解与生成领域的核心研究方向。关于具体型号的技术细节，目前官方尚未全面公开，但可以分享一些通用的技术背景：

### 模型原理概述
1. **多模态融合架构**  
   - 采用跨模态注意力机制，能够联合建模文本、图像、视频等多种输入模态。
   - 核心设计包含：
     - 文本编码器（基于改进型Transformer架构）
     - 视觉特征提取模块（支持像素级语义分割及目标检测）
     - 跨模态对齐层（通过对比学习实现模态间表征空间的统一）

2. **训练方法论**
   - 三阶段训练流程：
     - 自监督预训练（百亿级图文对数据，掩码语言建模+图像复原任务）
     - 领域自适应微调（垂直领域专家数据注入）
     - 人类反馈强化学习（RLHF提升响应质量）

3. **轻量化设计特点**
   - 动态稀疏激活机制（仅激活相关子网络）
   - 知识蒸馏压缩（教师-学生模型架构）
   - 混合精度计算（FP16/INT8量化）

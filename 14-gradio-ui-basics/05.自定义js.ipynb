```python
import gradio as gr

with gr.Blocks() as demo:
    text = gr.Textbox(placeholder='请输入文本内容')
    radio = gr.Radio(['广州','深圳','北京','上海'])

    with gr.Row():
        btn1 = gr.Button('创建句子')

    output1 = gr.Textbox(label='输出的句子')
    output2 = gr.Textbox(label='结果进行翻转')

    def createSentence(text,radio):
        return f'{text}{radio}'
    btn1.click(createSentence,[text,radio],output1)
    jsCode='''
    (x)=>{
    document.querySelector('gradio-app').style.background = 'skyblue'
    return x.split('').reverse().join('')
    }
    '''

    output1.change(lambda x:x,output1,output2,js=jsCode)

demo.launch()
```

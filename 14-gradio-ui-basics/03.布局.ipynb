```python
import gradio as gr
def variable_outputs(k):
    k=int(k)
    return [gr.Textbox(visible=True)]*k+[gr.Textbox(visible=False)]*(10-k)

with gr.Blocks() as demo:
    s=gr.Slider(1,10,value=10,step=1,label='显示多少个结果')

    textboxes=[]
    for i in range(10):
        t=gr.Textbox(f'textbox{i}')
        textboxes.append(t)

    s.change(variable_outputs,s,textboxes)

demo.launch()
```

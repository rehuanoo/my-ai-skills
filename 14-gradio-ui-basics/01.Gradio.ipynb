```python
# 图像组件
import numpy as np
def greet(input_image):
    sepia_filter = np.array([
        [0.3,0.7,0.2],
        [0.2,0.6,0.1],
        [0.3,0.5,0.1]
    ])

    result = input_image.dot(sepia_filter.T)
    result = result/result.max()
    return result
    


demo = gr.Interface(fn=greet,inputs=gr.Image(),outputs='image')
print(__name__)

if __name__ =='__main__':
    print('启动')
    demo.launch()
```

```python
import gradio as gr
# 数据传递
def a1(A):
    return A

def a2(B):
    return B+10

with gr.Blocks() as demo:
    a = gr.Number(label='A')
    b = gr.Number(label='B')

    atob = gr.Button('AtoB')
    btoa = gr.Button('BtoA')
    
    atob.click(a1,inputs=a,outputs=b)
    btoa.click(a2,inputs=b,outputs=a)
if __name__ =='__main__':
    print('启动')
    demo.launch()
```

```python
def a1(x):
    return {
        c:x[a]+x[b],
        d:x[a]-x[b]
    }

with gr.Blocks() as demo:
    a = gr.Number(label='A')
    b = gr.Number(label='B')
    with gr.Row():
        atob = gr.Button('运算')
    c = gr.Number(label='相加结果')
    d = gr.Number(label='相减结果')

    
    
    
    atob.click(a1,inputs={a,b},outputs={c,d})
    
if __name__ =='__main__':
    print('启动')
    demo.launch()
```

```python
def change_textbox(choice):
    if choice == 'short':
        return gr.Textbox(lines=2,visible=True)
        # return gr.Button('duande')
    elif choice == 'long':
        return gr.Textbox(lines=8,visible=True)
    else :
        return gr.Textbox(lines=2,visible=False)

with gr.Blocks() as demo:
    radio = gr.Radio(
        ['short','long','none'],label='请选择输入长度'
    )
    text = gr.Textbox(lines=2,interactive=True)
    radio.change(
        fn=change_textbox,inputs=radio,outputs=text
    )
demo.launch()
```

```python
import random
import time

with gr.Blocks() as demo:
    chatbot = gr.Chatbot()
    msg = gr.Textbox()
    clear=gr.Button('Clear')
    def user_fn(user_msg,history):
        history.append({"role":"user", "content": user_msg})
        return "", history

    def bot_fn(history):
        bot_message = random.choice(['你好，主人！','请问有什么能够帮到你？','很开心与你交谈！'])
        time.sleep(2)
        history.append({"role":"assistant", "content": bot_message})
        return history

    msg.submit(user_fn,[msg,chatbot],[msg,chatbot]).then(
        bot_fn,chatbot,chatbot
    )
demo.launch()
```

```python
import gradio as gr
import random
import time
import math
import numpy as np
import plotly.express as px

plot_end = 2*math.pi

def get_plot(period=1):
    global plot_end
    x=np.arange(plot_end-2*math.pi,plot_end,0.02)
    y = np.sin(x*period*2*math.pi)
    fig = px.line(x=x,y=y)

    plot_end += 2*math.pi
    if plot_end>1000:
        plot_end = 2*math.pi
    return fig
    
with gr.Blocks() as demo:
    with gr.Row():
        with gr.Column():
            gr.Markdown('每秒跟新正弦曲线')
            period = gr.Slider(label='每周期图表',value=1,minimum=0,maximum=10,step=1)
            plot = gr.Plot(label='展示余弦曲线')
    dep = demo.load(get_plot,None,plot,every=1)
    period.change(get_plot,period,plot)
demo.launch()
```

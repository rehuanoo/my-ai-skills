```python
import gradio as gr

def calculator(num1,operation,num2):
    if operation =='add':
        return num1+num2
    elif operation =='sub':
        return num1-num2
    elif operation =='mul':
        return num1*num2
    elif operation =='div':
        if num2==0:
            raise gr.Error('除数不能为0')
        return num1/num2


demo=gr.Interface(
    calculator,
    [
        'number',
        gr.Radio(['add','sub','mul','div']),
        'number'
    ],
    'number',
    examples=[
        [45,'add',3],
        [3.14,'div',2],
        [7,'mul',5],
        [128,'sub',20]
    ]
)

demo.launch()
```

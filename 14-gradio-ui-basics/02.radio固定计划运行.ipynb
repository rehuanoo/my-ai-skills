```python
import gradio as gr
import random
import time
import math
import numpy as np
import plotly.express as px

plot_end = 2*math.pi

def get_plot(period=1):
    global plot_end
    x=np.arange(plot_end-2*math.pi,plot_end,0.02)
    y = np.sin(x*period*2*math.pi)
    fig = px.line(x=x,y=y)

    plot_end += 2*math.pi
    if plot_end>1000:
        plot_end = 2*math.pi
    return fig
    
with gr.Blocks() as demo:
    with gr.Row():
        with gr.Column():
            gr.Markdown('每秒跟新正弦曲线')
            period = gr.Slider(label='每周期图表',value=1,minimum=0,maximum=10,step=1)
            plot = gr.Plot(label='展示余弦曲线')
    # Gradio 5.x/6.x 移除了 every 参数，改用 gr.Timer 实现定时刷新
    demo.load(get_plot, period, plot)
    timer = gr.Timer(value=1)
    timer.tick(get_plot, period, plot)
    period.change(get_plot, period, plot)
demo.launch()
```

```python
import gradio as gr


with gr.Blocks() as demo:
    turn =gr.Textbox('X',interactive=False,label='Turn')
    board = gr.Dataframe(value=[['','','']]*3,interactive=False,type='array')

    def place(turn,board,evt:gr.SelectData):
        if evt.value:
            return turn,board

        board[evt.index[0]][evt.index[1]] = turn
        if turn =='X':
            turn='0'
        else:
            turn='X'
        return turn,board

    board.select(place,[turn,board],[turn,board],show_progress='hidden')
demo.launch()
```

```python
import random
import time

with gr.Blocks() as demo:
    chatbot = gr.Chatbot()
    msg = gr.Textbox()
    btn = gr.Button('提交')
    clear=gr.Button('Clear')
    def user_fn(user_msg,history):
        history.append({"role":"user", "content": user_msg})
        return "", history

    def bot_fn(history):
        bot_message = random.choice(['你好，主人！','请问有什么能够帮到你？','很开心与你交谈！'])
        time.sleep(2)
        history.append({"role":"assistant", "content": bot_message})
        return history

    # msg.submit(user_fn,[msg,chatbot],[msg,chatbot]).then(
    #     bot_fn,chatbot,chatbot
    # )
    gr.on(
        triggers = [msg.submit,btn.click],
        fn=user_fn,
        inputs=[msg,chatbot],
        outputs=[msg,chatbot]
    )
    gr.on(
        triggers = [msg.submit,btn.click],
        fn=bot_fn,
        inputs=chatbot,
        outputs=chatbot
    )
demo.launch()
```

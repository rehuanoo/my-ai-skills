```python
import numpy as np
a=['hello']
b=['world']
c = np.char.add(a,b)
c
```

```python
d=['Hello']
e=['World']
f = np.char.add(d,e)
np.char.lower(f)
```

```python
np.char.upper(f)
```

```python
# 字符串制定参数,字符串居中
np.char.center(
    ['hello','world','laochen'],
    width=20,
    fillchar='-'
)
```

```python
urlStr1='http://www.baidu.com/imgs/flower.jpg'
urlStr2='http://www.taobao.com/public/imgs/dog.jpg'
np.char.split([urlStr1,urlStr2],'/')
```

```python
# 按照换行符分割
article = 'laochen\n小明\n小红\n'
print(article)
```

```python
np.char.splitlines([article])
```

```python
# 移除开头结尾的空白符
a1='       \n        hello   laochen   \n      '
np.char.strip(a1)
```

```python
np.char.join([':'],['nihao'])
```

```python
# 字符串的替换
str1 = 'xm666nihao'
str2 = 'xm777nihao'
np.char.replace([str1,str2],'xm','[hxc]')
```

```python
# 编码
a = np.char.encode(['你好，我是胡星辰555'],'utf-8')
a
```

```python
# 解码
np.char.decode(a,'utf-8')
```

```python
n = [1,2,3,4,5,6,7]
np.power(n,8)
```

```python
np.square(n)
```

```python
a = np.array([1,2,3,4,5,6])
a**3
```

```python
# 获取最大值的索引值
np.argmax(a)
```

```python
# 保存文本
arr= np.arange(10).reshape(2,5)
np.savetxt('save-array.txt',arr)
```

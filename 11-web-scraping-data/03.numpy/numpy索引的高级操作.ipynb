```python
# 高级索引的方式
b = np.array([[1,2],[3,4],[5,6]])
b[[0,1,2],[0,1,0]]
# [0,1,2]只是第几个元素，[0,1,0]是指每个元素中的第几个
```

```python
# 布尔索引
price = np.array([7,8,9,10,11,12])
res = price>10
```

```python
res
```

```python
price[res]
```

```python
salesNum = np.array([1,2,3,4,5,6])
price[salesNum<4]
```

```python
# 多维数组的切片
a =np.array([[1,2,3],[4,0,3],[1,0,1]])
# a里面每个元素的第一个数
a[:,0]
```

```python
# 每个元素的第二第三个数
a[:,1:]
```

```python
# 第二第三个元素的第二第三个数
a[1:,1:]
```

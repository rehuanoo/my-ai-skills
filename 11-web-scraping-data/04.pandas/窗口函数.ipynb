```python
import pandas as pd
import numpy as np
data={
    'id':[1,5,8,9],
    'name':['小明','小红','小刚','小李'],
    'age':[15,18,16,13],
    'cid':[11,12,13,14]
}
df = pd.DataFrame(data)
df
```

```python
rollwin = df.rolling(window=2)
rollwin
```

```python
data1={
    'classname':['一年一班','一年二班','一年三班','一年四班'],
    'cid':[11,12,13,14]
}
df2 = pd.DataFrame(data1,index=[11,12,13,14])
df2
```

```python
pd.merge(df,df2,left_on='cid',right_index=True)
```

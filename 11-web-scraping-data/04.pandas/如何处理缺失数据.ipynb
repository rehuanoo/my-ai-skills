```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    'a':[1,5,np.nan,9],
    'b':[6,20,65,np.nan],
    'c':[5,np.nan,np.nan,30]
})
df
# None，pd.Na也是空值
```

```python
df.isnull()
```

```python
# 删除含有nan的行
df.dropna()
```

```python
# 删除含有nan的列
df.dropna(axis=1)
```

```python
# 缺失值填充
df.fillna(0)
```

```python
# 缺失值填均值
df.fillna(df.mean())
```

```python
# 缺失值填中位数
df.fillna(df.median())
```

```python
# 向前填充
df.ffill()
```

```python
# 向后填充
df.bfill()
```

```python
# 线性差值
# 空值不能用None，pd.NA
df.interpolate()
```

```python
import pandas as pd
df=pd.DataFrame({
    'a':[1,2,3],
    'b':[4,5,9]
})
df
```

```python
def range_diff(x):
    print(x)
    return x.max()-x.min()

# 按照每一行操作
col_range = df.apply(range_diff,axis=0)
col_range
```

```python
# 按照每一列执行
row_range = df.apply(range_diff,axis=1)
row_range
```
